```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

from flask import Flask, request, jsonify
from flask_cors import CORS
from pyngrok import ngrok
```

```python
np.random.seed(42)

n = 2000

stocks = np.random.uniform(0, 100, n)
bonds = np.random.uniform(0, 100 - stocks)

remaining = 100 - stocks - bonds
gold = remaining

volatility = (
    stocks * 0.30 +
    np.random.normal(5, 5, n)
)

volatility = np.clip(volatility, 2, 40)

df = pd.DataFrame({
    "stocks_percent": stocks,
    "bonds_percent": bonds,
    "gold_percent": gold,
    "volatility": volatility
})

df.head()
```

```python
def calculate_risk(row):

    score = (
        row["stocks_percent"] * 0.5 +
        row["volatility"] * 1.5 -
        row["bonds_percent"] * 0.15
    )

    if score < 35:
        return "Low"
    elif score < 65:
        return "Medium"
    else:
        return "High"


df["risk"] = df.apply(calculate_risk, axis=1)

df["risk"].value_counts()
```

```python
print(df.head(10))
print("\nDataset shape:", df.shape)
print("\nRisk distribution:")
print(df["risk"].value_counts())
```

```python
X = df[
    [
        "stocks_percent",
        "bonds_percent",
        "gold_percent",
        "volatility"
    ]
]

y = df["risk"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
logistic_model = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=1000))
])

logistic_model.fit(X_train, y_train)

logistic_pred = logistic_model.predict(X_test)

logistic_accuracy = accuracy_score(
    y_test,
    logistic_pred
)

print(
    "Logistic Regression Accuracy:",
    round(logistic_accuracy * 100, 2),
    "%"
)
```

```python
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)

rf_accuracy = accuracy_score(
    y_test,
    rf_pred
)

print(
    "Random Forest Accuracy:",
    round(rf_accuracy * 100, 2),
    "%"
)
```

```python
if rf_accuracy >= logistic_accuracy:
    best_model = rf_model
    best_model_name = "Random Forest"
    best_accuracy = rf_accuracy
else:
    best_model = logistic_model
    best_model_name = "Logistic Regression"
    best_accuracy = logistic_accuracy

print("Selected Model:", best_model_name)
print("Accuracy:", round(best_accuracy * 100, 2), "%")
```

```python
final_prediction = best_model.predict(X_test)

print(
    classification_report(
        y_test,
        final_prediction,
        zero_division=0
    )
)
```

```python
test_portfolio = pd.DataFrame([{
    "stocks_percent": 70,
    "bonds_percent": 20,
    "gold_percent": 10,
    "volatility": 25
}])

result = best_model.predict(test_portfolio)[0]

print("Predicted Risk:", result)
```

```python
app = Flask(__name__)
CORS(app)

@app.route("/", methods=["GET"])
def home():

    return jsonify({
        "status": "success",
        "message": "Portfolio Risk Classifier API is running",
        "model": best_model_name,
        "accuracy": round(best_accuracy * 100, 2)
    })
```

```python
@app.route("/predict", methods=["POST"])
def predict():

    try:

        data = request.get_json()

        required_fields = [
            "stocks_percent",
            "bonds_percent",
            "gold_percent",
            "volatility"
        ]

        for field in required_fields:
            if field not in data:
                return jsonify({
                    "error": f"Missing field: {field}"
                }), 400

        stocks = float(data["stocks_percent"])
        bonds = float(data["bonds_percent"])
        gold = float(data["gold_percent"])
        volatility = float(data["volatility"])

        if stocks < 0 or bonds < 0 or gold < 0:
            return jsonify({
                "error": "Percentages cannot be negative"
            }), 400

        total = stocks + bonds + gold

        if abs(total - 100) > 0.01:
            return jsonify({
                "error": "Stocks + Bonds + Gold must equal 100%"
            }), 400

        if volatility < 0:
            return jsonify({
                "error": "Volatility cannot be negative"
            }), 400

        portfolio = pd.DataFrame([{
            "stocks_percent": stocks,
            "bonds_percent": bonds,
            "gold_percent": gold,
            "volatility": volatility
        }])

        prediction = best_model.predict(portfolio)[0]

        return jsonify({
            "status": "success",
            "portfolio": {
                "stocks_percent": stocks,
                "bonds_percent": bonds,
                "gold_percent": gold,
                "volatility": volatility
            },
            "risk": prediction,
            "model": best_model_name
        })

    except Exception as e:

        return jsonify({
            "error": str(e)
        }), 400
```
